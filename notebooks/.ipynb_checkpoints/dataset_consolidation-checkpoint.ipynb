```python
import pandas as pd 
from pathlib import Path 
data_dir = Path('../data')
resume_dataset1=pd.read_csv(data_dir/'raw/resumes_dataset.csv')
resume_dataset2=pd.read_csv(data_dir/'raw/UpdatedResumeDataSet.csv')
```

```python
## 1. RESUME DATASET
```

### Lecture et affichage des donnees

```python
#Lecture et affichage Information dataframme
def read_data(data)->None:
    print("\n __Aperçu des données :")
    print(data.head(3))
    # Afficher les informations sur le DataFrame
    print("\n __Informations sur les données :")
    print(data.info())
    print(" \n __Taille : \n",data.shape)
    print("\n __Information sur les Categories: \n",data['Category'].value_counts().reset_index())
```

```python
read_data(resume_dataset1)
```

```python
read_data(resume_dataset2)
```

##### Suppression des lignes avec categories Biasees  dataset1

```python
# Liste des catégories à supprimer / recuperer en amont pendant la reuperation du dataset
categories_a_supprimer = [
    'Troy', 'Robert Half', 'Edison', 'Unknown','Chandler','Irving','Laurel','Charles Schwab'
]
# Filtrer pour garder seulement les lignes NON dans cette liste
new_resume_dataset1 = resume_dataset1[~resume_dataset1['Category'].isin(categories_a_supprimer)]
read_data(new_resume_dataset1)
```

#### Verification Valeur manquante et chaine vide des dataset

```python
from numpy import empty


def missing_values(data1,data2=empty )->None:
    #Verification Valeur manquante
    print("\n  Dataseset1:")
    print(" \t Valeur manquante  ::")
    print(data1.isna().sum())
    print("\n Chaine vide:")
    print("Resume",data1['Resume'].apply(lambda x: str(x).strip() == '').sum())
    print("Category",data1['Category'].apply(lambda x: str(x).strip() == '').sum())
    
    if data2 is not empty :
        print(" \n \t -------------------------------:: \n ")
        print("  Dataseset2:")
        print(" \t Valeur manquante  ::")
        print(data2.isna().sum())
        print(" \t \t Chaine vide   ::")
        print("Resume",data2['Resume'].apply(lambda x: str(x).strip() == '').sum())
        print("Category",data2['Category'].apply(lambda x: str(x).strip() == '').sum())
    else:
        print(" \n ")
```

```python
missing_values(new_resume_dataset1,resume_dataset2)
```

### Combinaison des 2 dataset :resumes_dataset et UpdateResume

```python
combined_resume_dataset = pd.concat([new_resume_dataset1, resume_dataset2], axis=0, ignore_index=True)
# Vérification
print(f"Taille dataset1: {len(new_resume_dataset1)} lignes")
print(f"Taille dataset2: {len(resume_dataset2)} lignes")
print(f"Taille combinée: {len(combined_resume_dataset)} lignes")
print("\n Aperçu du résultat:")
print(combined_resume_dataset.head(3))
```

#### Verification Valeur manquante et chaine vide des dataset

```python
missing_values(combined_resume_dataset)
```

```python
read_data(combined_resume_dataset)
```

#### Suppresssion des doublons 

```python
combined_resume_dataset = combined_resume_dataset.drop_duplicates()
read_data(combined_resume_dataset)
```

### Combinaison DATASET FINAL

#### Ajout 3eme dataset resume_dataset_JitWah

```python
resume_dataset3=pd.read_csv(data_dir/'raw/combined_resume_dataset_JitWah.csv')
read_data(resume_dataset3)
```

#### Combinaison des dataset :resume_dataset3 et combined_resume_dataset

```python
combined_resume_dataset_final = pd.concat([combined_resume_dataset, resume_dataset3], axis=0, ignore_index=True)

# Vérification
print(f"Taille dataset1c: {len(combined_resume_dataset)} lignes")
print(f"Taille dataset3: {len(resume_dataset3)} lignes")
print(f"Taille combinée final: {len(combined_resume_dataset_final)} lignes")
read_data(combined_resume_dataset_final)
```

#### Suppresssion des doublons 

```python
combined_resume_dataset_final = combined_resume_dataset_final.drop_duplicates()
read_data(combined_resume_dataset_final)
```

```python
missing_values(combined_resume_dataset_final)
```

#### Enregistrement dataset

```python
# new_resume_dataset1.to_csv(data_dir/'interim/resumes_clean_Category.csv', index=False)
combined_resume_dataset_final.to_csv(data_dir/'interim/combined_resume_final.csv', index=False)
```

## 2. Job Offer DATASET

```python
offer_dataset1=pd.read_csv(data_dir/'raw/new_job_description.csv')
offer_dataset2=pd.read_csv(data_dir/'raw/Cleaned_DS_Jobs.csv')
```

```python
offer_dataset1.head()
```

```python
offer_dataset2.head()
```

#### Suppression colonnes

```python
columns_to_drop=['Rating','Headquarters','Location','Company Name','Size',
                 'Type of ownership','Salary Estimate','Industry','python','excel',
                'hadoop','spark','aws','tableau','big_data','job_simp','seniority',
                'Sector','Revenue','min_salary','max_salary','avg_salary','job_state','same_state','company_age']


offer_dataset2=offer_dataset2.drop(columns=columns_to_drop)
offer_dataset2=offer_dataset2.rename(columns={'Job Title':'Job Title_Category'})
offer_dataset2.head()
```

#### Verification Valeur manquante

```python

print("\n  Dataseset1:")
print(" \t Valeur manquante  ::")
print(offer_dataset1.isna().sum())
print("\n Chaine vide:")
print("Job Description",offer_dataset1['Job Description'].apply(lambda x: str(x).strip() == '').sum())
print("Job Title_Category",offer_dataset1['Job Title_Category'].apply(lambda x: str(x).strip() == '').sum())
print(" \n \t -------------------------------:: \n ")
print("  Dataseset2:")
print(" \t Valeur manquante  ::")
print(offer_dataset2.isna().sum())
print(" \t \t Chaine vide   ::")
print("Job Description",offer_dataset2['Job Description'].apply(lambda x: str(x).strip() == '').sum())
print("Job Title_Category",offer_dataset2['Job Title_Category'].apply(lambda x: str(x).strip() == '').sum())
```

### Combinaison des DATASET

```python
combined_offer_dataset = pd.concat([offer_dataset1, offer_dataset2], axis=0, ignore_index=True)
combined_offer_dataset =combined_offer_dataset.drop_duplicates()
combined_offer_dataset =combined_offer_dataset.dropna()
# Vérification
print(f"Taille dataset1: {len(offer_dataset1)} lignes")
print(f"Taille dataset2: {len(offer_dataset2)} lignes")
print(f"Taille combinée: {len(combined_offer_dataset)} lignes")
print("\n Aperçu du résultat:")
print(combined_offer_dataset.head(3))
```

```python
combined_offer_dataset.to_csv(data_dir/'interim/combined_offer_dataset.csv', index=False)
```
