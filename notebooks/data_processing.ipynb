```python
import pandas as pd 
from pathlib import Path 
data_dir = Path('../data')
resume_dataset1=pd.read_csv(data_dir/'raw/resumes_dataset.csv')
resume_dataset2=pd.read_csv(data_dir/'raw/UpdatedResumeDataSet.csv')
```

### Lecture et affichage des donnees

```python
resume_dataset1.head()
```

```python
resume_dataset2.head()
```

```python
#Lecture et affichage Information dataframme
def read_data(data):
    print("\n __Aperçu des données :")
    print(data.head(3))
    # Afficher les informations sur le DataFrame
    print("\n __Informations sur les données :")
    print(data.info())
    print(" \n __Taille : \n",data.shape)
    print("\n __Information sur les Categories: \n",data['Category'].value_counts().reset_index())
```

```python
read_data(resume_dataset1)
```

```python
read_data(resume_dataset2)
```

##### Suppression des lignes avec categories Biasees  dataset1

```python
# Liste des catégories à supprimer / recuperer en amont pendant la reuperation du dataset
categories_a_supprimer = [
    'Troy', 'Robert Half', 'Edison', 'Unknown','Chandler','Irving','Laurel','Charles Schwab'
]
# Filtrer pour garder seulement les lignes NON dans cette liste
new_resume_dataset1 = resume_dataset1[~resume_dataset1['Category'].isin(categories_a_supprimer)]
read_data(new_resume_dataset1)
```

#### Verification Valeur manquante#####  et chaine vide dataset

```python
#Verification Valeur manquante
print(" \t Valeur manquante  ::")
print("\n _Dataset1:")
print(new_resume_dataset1.isna().sum())
print("\n Dataset2:")
print(resume_dataset2.isna().sum())
```

```python
#Verification chaine vide dataset
print(" \t \t Chaine vide   ::")
print("  Dataseset1:")

print("Resume",resume_dataset2['Resume'].apply(lambda x: str(x).strip() == '').sum())
print("Category",resume_dataset2['Category'].apply(lambda x: str(x).strip() == '').sum())
print("\n  Dataseset1:")
print("Resume",new_resume_dataset1['Resume'].apply(lambda x: str(x).strip() == '').sum())
print("Category",new_resume_dataset1['Category'].apply(lambda x: str(x).strip() == '').sum())
```

### Concactenation 2 dataset :resumes_dataset et UpdateResume

```python
combined_resume_dataset = pd.concat([new_resume_dataset1, resume_dataset2], axis=0, ignore_index=True)

# Vérification
print(f"Taille dataset1: {len(new_resume_dataset1)} lignes")
print(f"Taille dataset2: {len(resume_dataset2)} lignes")
print(f"Taille combinée: {len(combined_resume_dataset)} lignes")
print("\n Aperçu du résultat:")
print(combined_resume_dataset.head())
```

```python
print(" \t \t Chaine vide   ::")
print("  Dataseset1:")

print("Resume",combined_resume_dataset['Resume'].apply(lambda x: str(x).strip() == '').sum())
print("Category",combined_resume_dataset['Category'].apply(lambda x: str(x).strip() == '').sum())
```

```python
print(" \t Valeur manquante  ::")
print("\n _Dataset1:")
print(combined_resume_dataset.isna().sum())
print("\n Dataset2:")
print(combined_resume_dataset.isna().sum())
```

```python
read_data(combined_resume_dataset)
```

####Suppresssion des doublons 

```python
combined_resume_dataset = combined_resume_dataset.drop_duplicates()
read_data(combined_resume_dataset)
```

### DATASET FINAL

```python
combined_resume_dataset_final = pd.concat([combined_resume_dataset, resume_dataset3], axis=0, ignore_index=True)

# Vérification
print(f"Taille dataset1c: {len(combined_resume_dataset)} lignes")
print(f"Taille dataset3: {len(resume_dataset3)} lignes")
print(f"Taille combinée final: {len(combined_resume_dataset_final)} lignes")
read_data(combined_resume_dataset_final)
```
